```python
#recursion
def fact(n):
    if n==1:
        return n
    else:
        return n*fact(n-1)
```

```python
fact(10)
```

```python
#armstrong
def count_digit(n):
    dividend=n
    sum=0
    while dividend>0:
        rem=dividend%10
        
        sum+=1
        dividend=dividend//10
    print(sum)
        
```

```python
def isarmstrong(n):
    dividend=n
    sum=0
    while dividend>0:
        quo=dividend//10
        
        sum+=quo**count_digit(n)
        dividend=dividend//10
    if sum==n:
        print("yes")
    else:
        print("no")
    
```

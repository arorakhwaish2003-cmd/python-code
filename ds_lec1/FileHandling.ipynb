```python
# create a new file using write
# if file do not exist- it creates new file
# if file exist - it creates new file by overriding another file
f=open("FDW.txt",'w')
f.write("this is a write demo\n")
f.write("S1\n")
f.close()
```

```python
# append mode--> ab jo bi write hofa file ke ene me add hoga
# new file ni banegi
f=open("FDW.txt",'a')
f.write("this is a append demo\n")
f.write("S2")
f.close()
```

```python
# r+--> isme overwrite hojaega
# read me cursor sbse upar hota h to vhi ajaega ye text
f=open("FDW.txt",'r+')
f.write("this is a r+ demo\n")
f.write("S2")
f.close()
```

```python
# with is used jab hum bade bade code likhte h to hum file close krna bhuljste h
with open("FDW.txt",'r') as f:
    print(f.read())
print("file is closed") # with ka use se hum indentation se bahar ae file khud
#close hojaegi
```

```python
l=["s1","s2","s3","s4"]
with open("asgs.txt","w") as f:
    for line in l:
        f.write(line+"\n")
```

```python
i=1
while i<=4:
    try:
        a=int(input())
        b=int(input())
        print(a/b)

    except ZeroDivisionError:
        print("wrong input b")
    except ValueError:
        print("wrong input value")
    except Exception:
        print("Error!!!!")
        
    i+=1
```

```python
#finally block
def fn(a,b,c):
    try:
        print(a)
        if b<0:
            raise Exception("B is negative")  #raise exception basically ek class banadi or 
            # b is negaative uska attribute h
        print(b)
    except:
        print(0)
    finally:
        print(c)

```

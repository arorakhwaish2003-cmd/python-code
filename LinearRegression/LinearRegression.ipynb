```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
```

```python
X=pd.read_csv('Linear_X_Train.csv').values
y=pd.read_csv('Linear_Y_Train.csv').values.reshape(-1)
print(X.shape,y.shape)
```

```python
plt.scatter(X,y)
plt.show()
```

```python
w=np.array([2])
w0=1

def hypothesis(X,w,w0):
    return w.dot(X.T)+w0
print(hypothesis(X,w,w0))
```

```python
def gradientDescent(X,y,lr=0.01,epochs=100):
    w=np.zeros(shape=(X.shape[1],))
    w0=1
    for e in range(epochs):
        ypred=hypothesis(X,w,w0)
        grads=np.zeros(shape=(w.shape[0],))


        for j in range(w.shape[0]):
            for i in range(X.shape[0]):
                grads[j]+=(ypred[i]-y[i])*X[i,j]
            grads[j]/=X.shape[0]
        w=w-lr*grads
        w0=w0-lr*np.mean(ypred-y)
    return w,w0
        
```

```python
w,w0=gradientDescent(X,y,lr=0.05,epochs=100)
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X=np.linspace(-4,2,100)
```

```python
Y=X**2+2*X+3
```

```python
plt.plot(X,Y)
plt.show()
```

```python
def dy_dx(X): #Gradient
# as wj ko replace x or y se
# x=x-aplha * dely/delx
    return 2*X+2
```

```python
alpha=0.2 # step size ki kitne step niche jana h
Xj=-4 # koi bi value leli
epochs=20 # kitne times eqn ko chalana h
for i in range(epochs):
    Xj=Xj-alpha*dy_dx(Xj)
    print(Xj)
    # dekho minimised Xj ki valyue -4 se leke -1 tk  jari h
    # or jo iska graph plot kia h usme bi minimum value -1 h
```

```python
# 20 times chala upr function
```

```python
# functn bana lete h
def fn(X):
    return X**2+2*X+3
Y=fn(X)
```

```python
alpha=0.2
Xs=[-4] # Xj ko Xs lelia
Xj=-4
epochs=20
for i in range(epochs):
    Xj=Xj-alpha*dy_dx(Xj)
    Xs.append(Xj)
```

```python
plt.plot(X,Y)
plt.plot(Xs,fn(np.array(Xs)),c='r')
plt.show()
```

```python
alpha=0.2
Xs=[2] # ksi bi direction me le skte h value
Xj=2
epochs=20
for i in range(epochs):
    Xj=Xj-alpha*dy_dx(Xj)
    Xs.append(Xj)
```

```python
plt.plot(X,Y)
plt.plot(Xs,fn(np.array(Xs)))
plt.show()
```

```python
plt.plot(X,Y) # ab zyda gap se niche jara h
plt.plot(Xs,fn(np.array(Xs)))
plt.show()
```

```python
N=int(input())
n=2
while n<=10:
    quo=N%n
    rem=N//n
    if rem==0:
        dividend=quo
        rem=dividend%n
    print(n)
    n+=1
```

```python
n=int(input())
dividend=n
sum=0
place=1
while dividend>0:
    rem=dividend%10
    quo=dividend//10
    sum+=10**(rem-1)*place
    dividend=quo
    place+=1
print(sum)
```

```python
N=int(input())
divisor=2
dividend=N
sum=0
sum1=0
while dividend>1:
    if dividend%divisor==0:
        factor=divisor
        while factor>0:
            sum+=factor%10
            factor//=10
        dividend//=10
    else:
        divisor+=1

while N>0:
    rem=N%10
    sum1+=rem
    N=N//10

if sum==sum1:
    print(1)
else:
    print(0)
```

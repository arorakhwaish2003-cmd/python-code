```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = pd.read_csv("xdata.csv",index_col=False,usecols=[1,2])
y = pd.read_csv("ydata.csv",index_col=False,usecols=[1])
# index col = False is not necessary ye hum islie likhte h ki index col ko by default lelo me ni dera 
# kyuki dataset me index col hai ni to khra h bydefault sare dal do
X_data,y_data = x.values,y.values.reshape(-1)
print(X_data.shape,y_data.shape)
```

```python
plt.scatter(X_data[y_data == 0][:,0],X_data[y_data == 0][:,1],c='blue',label = '0')
# x or y ki value lo jisme y 0 hai or phle sari rowd 1st col 
# then sari rows or 2nd col
plt.scatter(X_data[y_data == 1][:,0],X_data[y_data == 1][:,1],c='y',label = '1')
plt.scatter(2,3,c='r')
plt.legend()
plt.show()
```

```python
T=np.array([2,3])
T
```

```python
def distance(a,b):
    return np.sqrt(np.sum((a-b)**2))
    # a=[1,2,3]
    # b=[3,4,5]
    # element wise sub krna h a-b krdo
    # then square krdo sbka
    # sbko add krdo kyuki 1d 2d 3d kuch bi ho skta h
    # then sqrt kr do
```

```python
# KNN CODE
k=5
def predict_1(t): # t is prediction pt.
    dist=[]
    for p in X_data:
        dist.append(distance(t,p))

    knear=np.argsort(dist)[:k] # sort krde asc me
    # the first k values dede
    klabels=[]
    for p in knear:
        klabels.append(y_data[p])
        # ab dist nikal lie first 5 then labeling krni h
        # ki konse grp ko belong krra h data point
    labels,counts= np.unique(klabels,return_counts=True) # unique values dede label ki
    return labels[np.argmax(counts)]  # then unme se max value
```

```python
predict_1(T)
```

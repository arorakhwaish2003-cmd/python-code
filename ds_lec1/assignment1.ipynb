```python
T = int(input())  # Number of test cases

s = 1  # Test case counter

while s <= T:  # Loop for each test case
    M = int(input())  # Limit for Harshit
    N = int(input())  # Limit for Ayush

    sum1 = 0  # Sum for Harshit
    sum2 = 0  # Sum for Ayush
    m = 1  # Odd number for Harshit
    n = 2  # Even number for Ayush

    while True:  # Alternate between Harshit and Ayush
        # Harshit's turn
        sum1 += m
        if sum1 > M:
            print("Harshit")
            break
        m += 2  # Next odd number

        # Ayush's turn
        sum2 += n
        if sum2 > N:
            print("Ayush")
            break
        n += 2  # Next even number

    s += 1  # Move to the next test case
```

```python
N = int(input())  # Number of test cases
n = 1

while n <= N:
    sum1 = 0  # Sum of odd digits
    sum2 = 0  # Sum of even digits
    num = int(input())  # Input number
    dividend = num

    while dividend > 0:
        rem = dividend % 10  # Extract the last digit
        if rem in [1, 3, 5, 7, 9]:  # Check if the digit is odd
            sum1 += rem
        elif rem in [2, 4, 6, 8, 0]:  # Check if the digit is even
            sum2 += rem
        dividend //= 10  # Remove the last digit

    if sum1 % 3 == 0:  # Check if the sum of odd digits is divisible by 3
        print('yes')
    elif sum2 % 4 == 0:  # Check if the sum of even digits is divisible by 4
        print('yes')
    else:
        print('no')

    n += 1
```

```python
N = int(input())

# Prime factorization and sum of prime factor digits
divisor = 2
dividend = N
prime_factor_digit_sum = 0

while dividend > 1:
    if dividend % divisor == 0:
        # Calculate the sum of digits of the prime factor
        factor = divisor
        while factor > 0:
            prime_factor_digit_sum += factor % 10
            factor //= 10
        dividend //= divisor
    else:
        divisor += 1

# Sum of the digits of the original number
original_digit_sum = 0
temp = N
while temp > 0:
    original_digit_sum += temp % 10
    temp //= 10

# Check if the two sums are equal
if prime_factor_digit_sum == original_digit_sum:
    print(1)
else:
    print(0)
```

```python
n = int(input())
arr = [int(input()) for _ in range(n)]

# Edge case: 1 or 0 element
if n <= 1:
    print("true")
    exit()

i = 1

# First strictly decreasing part
while i < n and arr[i] < arr[i - 1]:
    i += 1

# Then strictly increasing part
while i < n and arr[i] > arr[i - 1]:
    i += 1

# If we used the full array and no invalid equal or zigzag
is_valid = True
for j in range(1, n):
    if arr[j] == arr[j - 1]:
        is_valid = False
        break

# Final result
if i == n and is_valid:
    print("true")
else:
    print("false")
```

```python
n = int(input())
words = input().split()

for i in range(n):
    if words[i] != "0":  # Check if already used
        print(words[i], end=" ")
        for j in range(i+1, n):
            if words[j] != "0" and sorted(words[i]) == sorted(words[j]):
                print(words[j], end=" ")
                words[j] = "0"  # Mark as used
        print()
```

```python
n = int(input())
words = input().split()

result = []

for i in range(n):
    if words[i] != "0":  # Check if already used
        group = [words[i]]
        for j in range(i+1, n):
            if words[j] != "0" and sorted(words[i]) == sorted(words[j]):
                group.append(words[j])
                words[j] = "0"  # Mark as used
        words[i] = "0"  # Mark current word as used
        result.append(group)

print(result)
```
